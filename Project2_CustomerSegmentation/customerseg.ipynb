```python
import pandas as pd
```

```python
df=pd.read_csv("ifood_df.csv")
```

```python
print("Missing values per column:")
print(df.isnull().sum())
```

```python
duplicatecount=df.duplicated().sum()
print(f"\nNumber of Duplicate rows found:{duplicatecount}")
```

```python
df = df.dropna()
```

```python
if duplicatecount>0:
    df=df.drop_duplicates()
    print("Duplicates Removed")
```

```python
df_final=df.copy()
df_final["C-Id"]=range(1,len(df_final)+1)#cId added for tableau dashboard
```

```python
#Data cleaned now Moving Forward to Tableau
df_final.to_csv("ready_for_tableau_project2.csv",index=False)
```
