```python
import pandas as pd
```

```python
df=pd.read_csv("AB_NYC_2019.csv")
```

```python

print(f"Duplicates Found in the Csv File:{df.duplicated().sum()}")
if(df.duplicated().sum()>0):
    print("Duplicates Found removing them")
    df=df.drop_duplicates()
else:
    print("Nothing is Perfomed As Zero Duplicates")
```

```python
print("Missing values per column:\n",df.isnull().sum())
```

```python
#In above observation we got to know there is a ghost column Unnamed:16 we will drop it
df.drop(columns=["Unnamed: 16"],inplace=True)
```

```python
# Some Names and Hostname are missing we will fill it with Unknown 
df["name"]=df["name"].fillna("Unknown")
df["host_name"]=df["host_name"].fillna("Unknown")
```

```python
df["reviews_per_month"]=df["reviews_per_month"].fillna(0)
df["last_review"]=df["last_review"].fillna("Never")
```

```python
# Now for the double checking 
print(df.isnull().sum())
```

```python
zeroprice=(df["price"]==0).sum()
print(f"There Are {zeroprice} zero price listing found")
```

```python
# Check how many rows have a price of 0
zero_prices = (df['price'] == 0).sum()
print(f"Number of zero price listings: {zero_prices}")

# Remove rows where price is 0
df = df[df['price'] > 0]

# Check the minimum price now (should be at least 1)
print(f"New minimum price: {df['price'].min()}")

```

```python
print(f"Missing values:\n{df.isnull().sum()}")
print(f"\n Final Shape Of The DataSet:{df.shape}")
```

```python
df.to_csv("Project3_Cleaned.csv",index=False)
print("Cleaned Successfully")
```
